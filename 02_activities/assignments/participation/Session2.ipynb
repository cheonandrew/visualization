```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Reds", fill=True)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length,
            cmap="Blues", fill=True, bw_adjust=0.5)
plt.show()
```

"The Iris dataset is a widely used dataset in machine learning, it contains information about different species of iris flowers. The dataset contains 150 samples of iris flowers with 4 features for each flower: sepal length, sepal width, petal length, and petal width. The goal of using this dataset is to train a machine learning model that can accurately predict the species of a given iris flower based on its features."


Give the aesthetic, substantive, and perceptual qualities of data visualizations, this `KDE plot` clearly depics the probability density of two continuous variables: *sepal width* and *sepal length* across 150 different samples of iris flowers. Frequency is clearly indicated by color intensity, variance is easily discriminable, and trends are immediately visible. It is effectively a smoother variation of a histogram.
