```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Set global font style
plt.rcParams.update({
    "axes.titleweight": "bold",
    "axes.labelweight": "bold",
    "font.family": "Arial",
    "font.size": 16,
    "axes.titlesize": 16,
    "axes.labelsize": 16,
    "xtick.labelsize": 16,
    "ytick.labelsize": 16,
    "legend.fontsize": 16
})
```

```python
# Convert "Hospitalizations by vaccination status" `.csv` data into `df`.
df = pd.read_csv("vac_status_hosp_icu.csv")
df["date"] = pd.to_datetime(df["date"])

# Use a high-contrast colormap
cmap = plt.get_cmap("Dark2")

# Prepare the figure axes
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the values for all three vaccination condition groups
ax.plot(
    df["date"],
    df["icu_unvac"],
    label="Unvaccinated",
    linewidth=2.25,
    color=cmap(6)
)

ax.plot(
    df["date"],
    df["icu_partial_vac"],
    label="Partially vaccinated",
    linewidth=2.25,
    color=cmap(0)
)

ax.plot(
    df["date"],
    df["icu_full_vac"],
    label="Fully vaccinated",
    linewidth=2.25,
    color=cmap(2)
)

# Stylize
ax.set_title("Ontario's COVID-19 ICU patients by vaccination status")
ax.set_xlabel("Date")
ax.set_ylabel("ICU patients")

legend = ax.legend(frameon=True)
legend.get_frame().set_edgecolor("black")
legend.get_frame().set_linewidth(1)

fig.autofmt_xdate()
plt.tight_layout()
plt.savefig("figure1.png", dpi=300)
plt.show()
```
